```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class QuardState(TypedDict):
    a: int
    b: int
    c: int
    equation: str
    discriment: float
    result: str
```

```python
def show_equation(state: QuardState):
    equation = f'{state["a"]}x^2 + {state["b"]}x + {state["c"]} = 0'
    return {'equation': equation}
```

```python
def calculate_discriment(state: QuardState):
    discriment = state["b"]**2 - (4*state["a"]*state["c"])
    return {'discriment': discriment}
```

```python
def real_roots(state: QuardState):
    root1 = (-state["b"] + (state["discriment"]**0.5)) / (2*state["a"])
    root2 = (-state["b"] - (state["discriment"]**0.5)) / (2*state["a"])
    return {'result': f'The roots are {root1} and {root2}'}
```

```python
def repeated_root(state: QuardState):
    root = -state["b"] / (2*state["a"])
    return {'result': f'The repeated root is {root}'}
```

```python
def no_real_roots(state: QuardState):
    return {'result': 'The equation has no real roots'}
```

```python
from typing import Literal


def check_condtion(state: QuardState)-> Literal["real_roots", "repeated_root", "no_real_roots"]:
    if state["discriment"] > 0:
        return "real_roots"
    elif state["discriment"] == 0:
        return "repeated_root"
    else:
        return "no_real_roots"
```

```python
graph = StateGraph(QuardState)
graph.add_node('show_equation',show_equation)
graph.add_node('calculate_discriment',calculate_discriment)


graph.add_node('real_roots',real_roots)
graph.add_node('repeated_root',repeated_root)
graph.add_node('no_real_roots',no_real_roots)

graph.add_edge(START,'show_equation')
graph.add_edge('show_equation','calculate_discriment')
# that's we can add condition and this check_condition is just a normal python function that takes the state and returns the next node name based on the condition.
graph.add_conditional_edges('calculate_discriment', check_condtion)



graph.add_edge('real_roots',END)
graph.add_edge('repeated_root',END)
graph.add_edge('no_real_roots',END)
work_flow =graph.compile()
```

```python
initial_state = {
    'a': 4,
    'b': -5,
    'c': -4}
```

```python
work_flow.invoke(initial_state)
```
