```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class Batsmanstats(TypedDict):
    runs:int
    balls:int
    fours: int
    sixes:int

    sr:float
    bpb : float
    boundary_percent: float
    summary: str
```

```python
def strikerate(state:Batsmanstats)-> Batsmanstats:
    sr = state['runs']/state['balls']*100
    # state[sr] =sr
    return {'sr': sr}
```

```python
def bpb(state:Batsmanstats)-> Batsmanstats:
    bpb = state['balls']/(state['fours']+state['sixes'])
    # state['bpb'] = bpb
    return {'bpb': bpb}
    # return state
```

```python
def boundary_percent(state:Batsmanstats)-> Batsmanstats:
    boundary_percent = ((state['fours']*4)+(state['sixes']*6))/state['balls']*100
    # state['boundary_percent'] = boundary_percent

    # return state
    return {'boundary_percent': boundary_percent}
```

```python
def summary(state:Batsmanstats)-> Batsmanstats:

    summary = f"Runs: {state['runs']}, Balls: {state['balls']}, Fours: {state['fours']}, Sixes: {state['sixes']}, SR: {state['sr']}, BPB: {state['bpb']}, Boundary Percent: {state['boundary_percent']}"
    # return state
    return {'summary': summary}
    
```

```python
graph = StateGraph(Batsmanstats)

# define node
graph.add_node("strikerate",strikerate)
graph.add_node("bpb",bpb)
graph.add_node("boundary_percent",boundary_percent)

graph.add_node("summary",summary)
# define edge
graph.add_edge(START,'strikerate')
graph.add_edge(START,'bpb')
graph.add_edge(START,'boundary_percent')
            

graph.add_edge('strikerate','summary')
graph.add_edge('bpb','summary')
graph.add_edge('boundary_percent','summary')

graph.add_edge('summary',END)
# compile graph
work_flow = graph.compile()
```

```python
initial_state ={
    'runs': 100,
    'balls': 50,
    'fours': 10,
    'sixes': 5,
}
```

```python
work_flow.invoke(initial_state)
```
